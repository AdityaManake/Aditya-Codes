```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

data = {
    'Experience': [1,2,3,4,5,6,7,8,9],
    'Salary': [45000,50000,60000,80000,110000,150000,200000,300000,500000]
}

df = pd.DataFrame(data)
X = df[['Experience']]
y = df['Salary']

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

pred_years = {"Experience": [11, 12, 13]}
pred_salaries = model.predict(pd.DataFrame(pred_years))

print("Predicted salaries for 11, 12, 13 years of experience:", pred_salaries)

mse = mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Squared Error:", mse)
print("Mean Absolute Error:", mae)
print("R-squared:", r2)
plt.scatter(X, y, color='blue')
plt.plot(X, y_pred, color='red')
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.show()
```
